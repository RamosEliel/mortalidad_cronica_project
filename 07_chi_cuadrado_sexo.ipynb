```python
import pandas as pd 
import matplotlib.pyplot as plt

df_mortalidad = pd.read_parquet("mortalidad_limpio.parquet")

df_mortalidad = df_mortalidad[df_mortalidad['ANO'] < 2026]
```

```python
from scipy.stats import chi2_contingency

# Tabla cruzada: filas=sexo, columnas=causa, valores=conteo
tabla_sexo = pd.crosstab(df_mortalidad['SEXO'], df_mortalidad['CLASIFICACION_CRONICAS'])
print(tabla_sexo)

# Prueba chi-cuadrado: compara si hay asociación entre sexo y causa
chi2_sexo, p_valor_sexo, gl_sexo, esperado_sexo = chi2_contingency(tabla_sexo)
print(f"Chi2: {chi2_sexo:.2f}, p-valor: {p_valor_sexo:.6f}, grados de libertad: {gl_sexo}")

# Misma tabla pero en porcentajes por fila (normalize='index'), para comparar proporciones entre sexos
tabla_prop = pd.crosstab(df_mortalidad['SEXO'], df_mortalidad['CLASIFICACION_CRONICAS'], normalize='index') * 100
print(tabla_prop.round(1))
```
