```python
import pandas as pd

url = "https://datosabiertos.bogota.gov.co/dataset/9fc1f14b-540b-4adc-93b1-ecd22780d6d2/resource/f33d3941-9030-4899-9420-42ef8757607b/download/mortalidad-prematura-por-enfermedades-cronicas-general.csv"
df_mortalidad = pd.read_csv(url, encoding='cp1252', sep=None, engine='python')
```

```python
def limpiar_edad(edad):
    if edad > 120:  # códigos con prefijo
        return int(str(edad)[-2:])
    return edad

df_mortalidad['EDAD_FALLECIDO'] = df_mortalidad['EDAD_FALLECIDO'].apply(limpiar_edad)
print(df_mortalidad['EDAD_FALLECIDO'].describe())
```

```python
print(df_mortalidad['SEXO'].value_counts())
print(df_mortalidad['NACIONALIDAD DE LOS CASOS'].value_counts())
print(df_mortalidad['REGIMEN_SEGURIDAD_SOCIAL'].value_counts())
print(df_mortalidad['EDAD_FALLECIDO'].value_counts())
print(df_mortalidad[(df_mortalidad['EDAD_FALLECIDO'] < 30) | df_mortalidad['EDAD_FALLECIDO'] > 70].shape[0])  
```

```python
df_mortalidad['REGIMEN_SEGURIDAD_SOCIAL'] = df_mortalidad['REGIMEN_SEGURIDAD_SOCIAL'].str.replace('¢', 'ó')
df_mortalidad['NACIONALIDAD DE LOS CASOS'] = df_mortalidad['NACIONALIDAD DE LOS CASOS'].str.replace('¢', 'ó')

print(df_mortalidad['REGIMEN_SEGURIDAD_SOCIAL'].value_counts())
```

```python
df_mortalidad.to_parquet('mortalidad_limpio.parquet', index=False)
```
